```python
# ============================================================
# chatbot_demo.png GENERATOR (Google Colab)
# Creates a professional chatbot demo image for poster
# Saves to: /content/drive/MyDrive/SurvivalVR_Outputs/chatbot_demo.png
# ============================================================

import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyBboxPatch
import os
import shutil

# ----------------------------
# 1. MOUNT DRIVE
# ----------------------------
try:
    from google.colab import drive
    drive.mount('/content/drive')
    DRIVE_PATH = "/content/drive/MyDrive/"
    print("Drive mounted successfully!")
except:
    DRIVE_PATH = "./"
    print("Saving locally.")

OUTPUT_FOLDER = os.path.join(DRIVE_PATH, "SurvivalVR_Outputs")
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
print(f"Output folder: {OUTPUT_FOLDER}\n")

# ----------------------------
# 2. CREATE CHATBOT DEMO IMAGE
# ----------------------------
fig, ax = plt.subplots(figsize=(12, 10))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Title
ax.text(5, 9.5, "SurvivalVR-BD Chatbot Demo", fontsize=18, fontweight='bold',
        ha='center', va='center', color='#0a2a44')

# Subtitle
ax.text(5, 9.0, "Rule-Based Survival Assistant for Coastal Fishermen", fontsize=12,
        ha='center', va='center', color='#2c3e50', style='italic')

# Chat bubbles with different colors
conversations = [
    ("User", "What's the weather?", '#D5E8D4', '#2E7D32', 8.0),
    ("Bot", "Check clouds. Dark cumulonimbus = storm within 6 hrs.\nCloud color = cyclone approaching. Wind > 60 knots = seek shelter!", '#E3F2FD', '#0D47A1', 6.5),
    ("User", "I need emergency help!", '#D5E8D4', '#2E7D32', 5.2),
    ("Bot", "EMERGENCY PROTOCOL:\n• Secure floating objects\n• Send bright/reflective signals\n• Use rope to bring others onboard\n• If sinking, create raft from debris", '#FFF3E0', '#BF360C', 3.0),
    ("User", "How to catch fish?", '#D5E8D4', '#2E7D32', 1.6),
    ("Bot", "Look for 10+ diving birds = fish school below.\nCurrent direction = fish location.\nDark green water = more fish.", '#E3F2FD', '#0D47A1', 0.0),
]

y_pos = 8.5
for speaker, text, bgcolor, textcolor, y in conversations:
    # Speaker label
    ax.text(1.2, y_pos + 0.25, speaker, fontsize=11, fontweight='bold', color=textcolor)

    # Bubble background
    bubble = FancyBboxPatch((1, y_pos - 0.5), 8, 0.8,
                            boxstyle="round,pad=0.1,rounding_size=0.2",
                            facecolor=bgcolor, edgecolor=textcolor, linewidth=1.5, alpha=0.9)
    ax.add_patch(bubble)

    # Text inside bubble
    ax.text(1.2, y_pos - 0.1, text, fontsize=10, color=textcolor,
            va='center', linespacing=1.3)

    y_pos -= 1.6

# Footer
ax.text(5, -1.2, "Powered by Rule Engine | IEEE WIE Day 2026",
        fontsize=10, ha='center', va='center', color='#7f8c8d')

plt.tight_layout()

# ----------------------------
# 3. SAVE IMAGE
# ----------------------------
local_path = "chatbot_demo.png"
plt.savefig(local_path, dpi=200, bbox_inches='tight', facecolor='white')
print(f"Image saved locally: {local_path}")

# Copy to Drive
drive_path = os.path.join(OUTPUT_FOLDER, "chatbot_demo.png")
shutil.copy(local_path, drive_path)
print(f"Image copied to Drive: {drive_path}")

# Show in Colab
try:
    from IPython.display import Image, display
    display(Image(local_path))
    print("\nImage displayed below:")
except:
    print("\nImage saved successfully. Open manually.")

# ============================================================
```
