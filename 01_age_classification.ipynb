# Cell 2: Create Project Folders in Drive

```python
# ============================================================
# CELL 2: Create Project Folders in Drive
# ============================================================
import os

# Base folder in Google Drive
BASE_DIR = '/content/drive/MyDrive/TeaGuardian/'

# Create all necessary folders
folders = [
    BASE_DIR,
    BASE_DIR + 'models/',
    BASE_DIR + 'notebooks/',
    BASE_DIR + 'plots/'
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"✅ Created: {folder}")

print("\n📁 All folders created successfully!")
```

# Cell 3: Upload Dataset Files

```python
# ============================================================
# CELL 3: Upload Dataset (Choose Files from Local Computer)
# ============================================================
from google.colab import files
import zipfile
import os

print("=" * 60)
print("📤 UPLOAD AGE QUALITY DATASET")
print("=" * 60)
print("Download from: https://data.mendeley.com/datasets/7t964jmmy3/1")
print()

# Upload the zip file
uploaded = files.upload()

# Get the filename
zip_filename = list(uploaded.keys())[0]
print(f"\n✅ Uploaded: {zip_filename}")

# Extract to the data folder
data_dir = '/content/age_data/'
os.makedirs(data_dir, exist_ok=True)

with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall(data_dir)

print(f"✅ Extracted to: {data_dir}")

# List extracted contents
print("\n📁 Extracted contents:")
print(os.listdir(data_dir))
```
