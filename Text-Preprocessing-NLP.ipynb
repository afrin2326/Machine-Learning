```python
!pip install pyspellchecker
import re
import nltk
import string
import unicodedata
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from spellchecker import SpellChecker
```

# Download required NLTK data

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
print("="*70)
print("(Raw Data with All Issues)")
print("="*70)

raw_data = """<html><body>
Hey Everyone! 😊 WELCOME to my NLP TUTORIAL...
Check out this awesome link: https://example.com/nlp-guide and also www.nlp101.com/blog

I'M so HAPPY to see U all here! 😍 This is GR8!!!
Did U know? NLP is rlly fascinating... LOL
Btw, I hv a question: "What's the difference between stemming & lemmatization?"
Pls let me know ur thoughts.

OMG!!! I cn't blve this is happening... 🎉💯
The movie was good but the ending was kinda boring...
U r awesome! ❤️🔥

Visit our website: http://www.amazing-nlp.com for more info!
Thx for reading. Have a gr8 day! Bye bye... 😊👍
</body></html>"""

print("\n[RAW DATA]:")
print(raw_data)
print("\n" + "="*70)
```

#   Step 1: Lowercasing

```python
print("\nSTEP 2: Lowercasing")
print("-"*50)

def lowercasing(text):
    return text.lower()

data = lowercasing(raw_data)
print("Output:")
print(data)
print("\n" + "="*70)
```

#  Step 2: Remove HTML tags

```python
print("\nSTEP 3: Remove HTML Tags")
print("-"*50)

def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return re.sub(pattern, '', text)

data = remove_html_tags(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 4: Remove URLs


```python
print("\nSTEP 4: Remove URLs")
print("-"*50)

def remove_urls(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return re.sub(pattern, '', text)

data = remove_urls(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 5: Remove Punctuation

```python
print("\nSTEP 5: Remove Punctuation")
print("-"*50)

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

data = remove_punctuation(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 6: Chat Word Treatment

```python
print("\nSTEP 6: Chat Word Treatment")
print("-"*50)

def chat_word_treatment(text):
    chat_words = {
        'u': 'you', 'ur': 'your', 'r': 'are', 'gr8': 'great',
        'lol': 'laugh out loud', 'brb': 'be right back',
        'gtg': 'got to go', 'idk': 'i do not know',
        'omg': 'oh my god', 'ttyl': 'talk to you later',
        'pls': 'please', 'plz': 'please', 'thx': 'thanks',
        'ty': 'thank you', 'yw': 'you are welcome',
        'np': 'no problem', 'b4': 'before', '2moro': 'tomorrow',
        '2day': 'today', 'nite': 'night', 'ppl': 'people',
        'hv': 'have', 'cn': 'can', 'blve': 'believe',
        'rlly': 'really', 'btw': 'by the way'
    }

    words = text.split()
    treated_words = [chat_words.get(word.lower(), word) for word in words]
    return ' '.join(treated_words)

data = chat_word_treatment(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 7: Spelling Correction

```python
print("\nSTEP 7: Spelling Correction")
print("-"*50)

spell = SpellChecker()

def spelling_correction(text):
    words = text.split()
    corrected_words = []

    for word in words:
        # Check if word is correct
        if word.lower() in spell:
            corrected_words.append(word)
        else:
            # Get correction for misspelled word
            corrected = spell.correction(word.lower())
            if corrected:
                corrected_words.append(corrected)
            else:
                corrected_words.append(word)

    return ' '.join(corrected_words)

data = spelling_correction(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 8: Removing Stopwords

```python
print("\nSTEP 8: Removing Stopwords")
print("-"*50)

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word.lower() not in stop_words]
    return ' '.join(filtered_words)

data = remove_stopwords(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 9: Handling Emojis

```python
print("\nSTEP 9: Handling Emojis")
print("-"*50)

def handle_emojis(text):
    # Convert emojis to text
    emoji_map = {
        '😊': 'smiley', '😂': 'laughing', '❤️': 'heart',
        '👍': 'thumbs up', '😢': 'crying', '😍': 'love',
        '🎉': 'celebration', '🔥': 'fire', '✨': 'sparkles',
        '💯': 'hundred'
    }

    for emoji, text_repr in emoji_map.items():
        text = text.replace(emoji, f' {text_repr} ')

    # Remove any remaining emojis
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags
                               "]+", flags=re.UNICODE)

    text = emoji_pattern.sub(r'', text)
    return text.strip()

data = handle_emojis(data)
print("Output:")
print(data)
print("\n" + "="*70)
```

# STEP 10: Tokenization

```python
print("\nSTEP 10: Tokenization")
print("-"*50)

def tokenization(text):
    return word_tokenize(text)

tokens = tokenization(data)
print("Output (Token List):")
print(tokens)
print(f"\nTotal tokens: {len(tokens)}")
print("\n" + "="*70)
```

# STEP 11: Stemming

```python
print("\nSTEP 11: Stemming")
print("-"*50)

stemmer = PorterStemmer()

def stemming(tokens):
    return [stemmer.stem(token) for token in tokens]

stemmed_tokens = stemming(tokens)
print("Output (Stemmed Tokens):")
print(stemmed_tokens)
print("\nStemming Examples:")
print(f"'running' → {stemmer.stem('running')}")
print(f"'studies' → {stemmer.stem('studies')}")
print(f"'better' → {stemmer.stem('better')}")
print("\n" + "="*70)
```

# STEP 12: Lemmatization

```python
print("\nSTEP 12: Lemmatization")
print("-"*50)

lemmatizer = WordNetLemmatizer()

def lemmatization(tokens):
    return [lemmatizer.lemmatize(token) for token in tokens]

lemmatized_tokens = lemmatization(tokens)
print("Output (Lemmatized Tokens):")
print(lemmatized_tokens)
print("\nLemmatization Examples:")
print(f"'running' → {lemmatizer.lemmatize('running')}")
print(f"'better' → {lemmatizer.lemmatize('better')}")
print(f"'geese' → {lemmatizer.lemmatize('geese')}")
print("\n" + "="*70)
```

# STEP 13: Stemming vs Lemmatization Comparison

```python
print("\nSTEP 13: Stemming vs Lemmatization Comparison")
print("="*70)

comparison_words = ['running', 'studies', 'better', 'geese', 'children', 'going']

print("\n{:15} {:15} {:15}".format("Original", "Stemmed", "Lemmatized"))
print("-"*45)

for word in comparison_words:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)
    print("{:15} {:15} {:15}".format(word, stemmed, lemmatized))

print("\n🔍 Key Differences:")
print("• Stemming: Cuts off word endings (fast but aggressive)")
print("• Lemmatization: Returns dictionary form (accurate but slower)")
```
